````python
# from image tensor to visual image
def back_to_image(img):
    # revert normalization
    # img = img / 2 + 0.5
    # npimg = img.numpy()
    # return np.transpose(npimg, (1, 2, 0))
    ###
    mean = torch.tensor([0.485, 0.456, 0.406]).view(3,1,1)
    std = torch.tensor([0.229, 0.224, 0.225]).view(3,1,1)

    img = img * std + mean
    img = img.clamp(0, 1)

    npimg = img.numpy()
    return np.transpose(npimg, (1, 2, 0))
````

## Why this model
The task requires the highest F1 score macro possible as goal and the dataset provided is not a big one.

It is also strongly suggested to use the library 'timm'.

Based on these factors, I needed a model that could generalyze well with a small/medium dataset and strong with transfer-learning.

After a few search, I tought a ConvNeXt model would have been the best options to fit the task between the ones timm provides.

Selecting indide the ConvNeXt family, I had to take in consideration also the computational limitations I would have faced so I finally choose the "tiny" version: ```convnext_tiny.fb_in22k_ft_in1k```.

````python
model = create_model("convnext_tiny.fb_in22k_ft_in1k", pretrained=True, num_classes=2)
````

## Fine Tuning and Hyperparameters

# Comments on the following choices:
Main point, why the following hyperparameters and fine tuning:

> batch_size = 128 (high for a small dataset), optimizer with values lr=1e-4, weight_decay=1e-4, layer_decay=0.75, epochs = 5 (small number) and no transfer learning.

The most relevant aspect that guided my decisions was the specific goal requested by the task:
> "Goal of the Project: The main goal is [...], able to classify the flowers autonomously with the best F1-score (macro) possible on test dataset".

So, regardless of the most "common" value selections (ex. small batch_size, etc..), I tried to optimize the values specificly to reach that goal.

This said, I tried different combinations of values (always keeping in mind the computational limitations of my environment) beginning with the most common choices like:

> batch_size in a range starting from 32 | epochs up to 20 | early stopping patience up to 5 | different values combinations for lr, weight_decay, layer_decay (ex. respectively: 1e-5 to 1e-3, 1e4 to 0.05, 0.65 to 0.85) | transfer learning freezing all layers despite the last one, etc..

but no one of the previous combinations reached the ```f1 score macro value of 0.9917``` achived by the combination I've finally choosen:

> batch_size = 128, epochs = 5, patience = 2, optimizer "adamw", lr=1e-4, weight_decay=1e-4, layer_decay=0.75, no transfer learning.

Another relevant point, as already mentioned, was related to the computational limitations of the environment; this made a good point to choose a model that reached optimal results with a low number of epochs.

````python
import torch.nn as nn
import torch
import torch.optim as optim
import torch.nn.functional as F
from torchsummary import summary
from sklearn.metrics import f1_score
````

````python
# model
x = torch.rand((1, 3, 224, 224))
logits = model(x)
if do_debug:
  print(logits.shape)
````

````python
criterion = nn.CrossEntropyLoss()
from timm.optim import create_optimizer_v2
optimizer = create_optimizer_v2(
    model,
    opt='adamw',
    # aggressive but best f1:
    lr=1e-4,
    weight_decay=1e-4,
    layer_decay=0.75
    # softer:
    # lr=1e-5,
    # weight_decay=0.05,
    # layer_decay=0.65
)
model.to(device);
````

````python
if do_debug:
  summary(model,(3,224,224))
````

````python
# NOTE: Best F1 macro is requested by the project task.
'''
Function to stop training if there's no improvement on F1 score macro
'''
class EarlyStopping:
  def __init__(self, save_path_spec, patience=5, min_delta=0):
    self.save_path = save_path_spec
    self.patience = patience
    self.min_delta = min_delta
    self.best_f1 = None
    self.counter = 0
    self.early_stop = False

  def __call__(self, f1, model):
    # manage first epoch
    if self.best_f1 is None:
      self.best_f1 = f1
      self.save_checkpoint(f1, model)
    # improved
    elif (f1 - self.best_f1) > self.min_delta:
      self.best_f1 = f1
      self.save_checkpoint(f1, model)
      self.counter = 0
    # not improved
    else:
      self.counter +=1
      if self.counter >= self.patience:
        self.early_stop = True
        print(f"[EARLY STOPPING] Best F1={self.best_f1:.4f} | Current F1={f1:.4f}")

  def save_checkpoint(self, f1, model):
      print(f"[CHECKPOINT] Saving best model at epoch={epoch} with f1_macro={f1:.4f}")
      # torch.save(model.state_dict(), self.save_path)
      torch.save({
          "epoch": epoch,
          "model_state_dict": model.state_dict(),
          "optimizer_state_dict": optimizer.state_dict(),
      }, self.save_path)
````

````python
# where to save if early stopping
convnext_tiny_save_folder = "models/convnext_tiny/"
os.makedirs(convnext_tiny_save_folder, exist_ok = True)
convnext_tiny_save_file = os.path.join(convnext_tiny_save_folder,"model.pt")
convnext_tiny_save_log = os.path.join(convnext_tiny_save_folder,"log.json")
````

````python
# Train the neural network
num_epochs = 5 #20
early_stopping = EarlyStopping(convnext_tiny_save_file, patience=2, min_delta=1e-4)
log_every_n_iter = 5
for epoch in range(num_epochs):
  # TRAIN
  model.train()
  epoch_loss = 0.0
  print(f"Epoch: {epoch}")
  for i, batch in enumerate(trainloader, 0):
    # batch --> (images, targets)
    x, targets = [el.to(device) for el in batch]
    # set optimizer at zero
    optimizer.zero_grad()
    # forward pass
    logits = model(x)
    # loss calculation
    loss = criterion(logits, targets)
    # backward pass
    loss.backward()
    # optimizer step --> update weights
    optimizer.step()
    epoch_loss += loss.item()
    if (i+1) % log_every_n_iter == 0:    # Print every log_every_n_iter mini-batches
        print(f"> iter [{i+1}/{len(trainloader)}] - train_loss={epoch_loss/log_every_n_iter:.4f}")
        epoch_loss = 0.0

  # VALIDATION
  model.eval()

  all_preds = []
  all_targets = []
  val_loss = []
  total, correct = 0, 0
  with torch.no_grad():
      for batch in valloader:
          x, targets = [el.to(device) for el in batch]
          logits = model(x)
          loss = criterion(logits, targets)
          val_loss.append(loss.item())
          preds = torch.argmax(logits, dim=1)

          all_preds.append(preds.cpu())
          all_targets.append(targets.cpu())

          total += targets.size(0)
          correct += (preds == targets).sum().item()

  all_preds = torch.cat(all_preds)
  all_targets = torch.cat(all_targets)

  total_f1 = f1_score(all_targets.numpy(), all_preds.numpy(), average="macro")
  print(total_f1)


  print(f"Validation report:\t validation_f1_macro={total_f1:.4f}")
  print(f"\t validation_loss={sum(val_loss)/len(val_loss):.4f} | validation_accuracy={correct/total:.4f}")


  #### EARLY STOPPING
  if early_stopping is not None:
      early_stopping(total_f1, model)
      if early_stopping.early_stop:
          print("Early stopping")
          break

  print(" \n *************** \n")

print('Finished Training')
````

# Validate best f1 model against Test dataset

````python
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
````

````python
# NOTE: The model used is the one saved (by EarlyStopping function) as best model.
checkpoint = torch.load(convnext_tiny_save_file)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

optimizer.load_state_dict(
    checkpoint["optimizer_state_dict"]
)

print(
    f"Loaded best model "
    f"from epoch {checkpoint['epoch']}"
)


# TEST EVALUATION
model.eval()

all_preds = []
all_targets = []
test_loss = []
total, correct = 0, 0
with torch.no_grad():
    for batch in testloader:
        x, targets = [el.to(device) for el in batch]

        logits = model(x)

        loss = criterion(logits, targets)
        test_loss.append(loss.item())

        _, preds = torch.max(logits, 1)

        all_preds.append(preds.cpu())
        all_targets.append(targets.cpu())

        total += targets.size(0)
        correct += (preds == targets).sum().item()

all_preds = torch.cat(all_preds)
all_targets = torch.cat(all_targets)

test_f1 = f1_score(all_targets.numpy(), all_preds.numpy(), average="macro")
print(test_f1)
#
print("FINAL TEST REPORT")
print(f"Test Loss: {sum(test_loss)/len(test_loss):.4f} | Test Accuracy: {correct/total:.4f} | Test F1 Macro: {test_f1:.4f}")

# classification report
print("\nClassification Report:")
print(classification_report(
    all_targets,
    all_preds,
    target_names=["Daisy", "Dandelion"]
))

# confusion matrix
cm = confusion_matrix(all_targets, all_preds)

plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt='d',
            xticklabels=["Daisy", "Dandelion"],
            yticklabels=["Daisy", "Dandelion"])

plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Confusion Matrix - Test Set")
plt.show()
````
